```python
import numpy as np

data = []

with open('input.txt', 'r') as file:
    for row in file.readlines():
        processed = []
        for ele in row[:-1]: # be careful of the goddarn '\n' end of line....
            if ele == '.':
                processed.append(0)
            else:
                processed.append(1)

        data.append(processed)

data = np.array(data)
```

```python
def less_than_4_neighbours(data: np.ndarray) -> int:
    # pad with zeros
    m, n = data.shape
    data = np.pad(data, [(1, 1), (1, 1)])
    kernel = np.ones((3, 3), dtype=bool)
    kernel[1, 1] = 0

    count = 0
    for i in range(1, m+1):
        for j in range(1, n+1):
            if data[i, j] == 1:
                count += data[i-1:i+2, j-1:j+2][kernel].sum() < 4

    return count

print(less_than_4_neighbours(data))

from scipy.signal import convolve2d
kernel = np.ones((3, 3), dtype=bool)
kernel[1, 1] = 0

def convolve_solution(data: np.ndarray) -> int:
    return np.where((convolve2d(data, kernel, mode='same', boundary='fill', fillvalue=0) < 4) & (data == 1))[0].size

convolve_solution(data)
```

```python
def maximum_removals(data: np.ndarray) -> int:
    # pad with zeros
    m, n = data.shape
    data = np.pad(data, [(1, 1), (1, 1)])
    kernel = np.ones((3, 3), dtype=bool)
    kernel[1, 1] = 0

    count = 0
    did_remove = True

    while did_remove:
        did_remove = False

        for i in range(1, m+1):
            for j in range(1, n+1):
                if data[i, j] == 1:
                    if data[i-1:i+2, j-1:j+2][kernel].sum() < 4:
                        did_remove = True
                        count += 1
                        data[i, j] = 0

    return count

maximum_removals(data)
```
