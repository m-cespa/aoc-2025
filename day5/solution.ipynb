```python
id_ranges = []
ids = []

process_ranges = True

with open('input.txt', 'r') as file:
    for line in file.readlines():
        if line == '\n':
            process_ranges = False
            continue

        if process_ranges:
            lower, upper = tuple(map(int, line[:-1].split('-')))

            id_ranges.append([lower, upper])
        else:
            ids.append(int(line[:-1]))

def merge_ranges(ranges: list[list[int]]) -> list[list[int]]:
    ranges.sort(key=lambda x: x[0])
    new_ranges = []

    n = len(ranges)
    merged = set()

    for i in range(n):
        if i in merged:
            continue
        curr_range = ranges[i][:] # copy

        if i == n-1:
            new_ranges.append(curr_range)
            break
        
        did_add = False
        for j in range(i+1, n):
            if not (curr_range[0] <= ranges[j][0] <= curr_range[1]):
                new_ranges.append(curr_range)
                did_add = True
                break
            else:
                curr_range[1] = max(curr_range[1], ranges[j][1])
                merged.add(j)

        if not did_add: # means the whole rest of the list got merged
            new_ranges.append(curr_range)
            break

    return new_ranges

id_ranges = merge_ranges(id_ranges)
```

```python
def valid_items() -> int:
    count = 0

    for id in ids:
        for id_range in id_ranges:
            if id_range[0] <= id <= id_range[1]:
                count += 1
                break

    return count

valid_items()
```

```python
def total_valid_items() -> int:
    count = 0

    for id_range in id_ranges:
        count += id_range[1] - id_range[0] + 1

    return count

total_valid_items()
```
