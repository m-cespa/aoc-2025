```python
p = 50
code = 0

with open("input.txt", "r") as file:
    for line in file.readlines():
        dir, step = line[0], int(line[1:])

        if dir == "L":
            new_p = p - step

            while new_p < 0:
                code += 1
                new_p += 100

        else:
            new_p = p + step

            while new_p > 99:
                code += 1
                new_p -= 100

        p = new_p

print(code)
```

```python
p = 50
code = 0

def method1(p: int, dir: str, steps: int) -> tuple[int, int]:
    zero_count = 0
    new_p = p

    step = -1 if dir == 'L' else 1
    a = -1 if dir == 'L' else 100
    b = 99 if dir == 'L' else 0

    for i in range(steps):
        if new_p == 0:
            zero_count += 1
        new_p += step
        if new_p == a:
            new_p = b

    return zero_count, new_p

def method2(p: int, dir: str, steps: int) -> int:
    zero_count = 0

    if dir == "L":
        new_p = p - steps

        while new_p < 0:
            zero_count += 1
            new_p += 100

    else:
        new_p = p + step

        while new_p > 99:
            zero_count += 1
            new_p -= 100

    return zero_count

with open("input.txt", "r") as file:
    for i, line in enumerate(file.readlines()):
        dir, steps = line[0], int(line[1:])

        zero_count_1, correct_p = method1(p, dir, steps)
        zero_count_2 = method2(p, dir, steps)

        p = correct_p
        code += zero_count_1

        if zero_count_1 != zero_count_2:
            print(f"Failure at line {i+1} with input {line}")
            print(f"Correct zero: {zero_count_1}, Incorrect zero: {zero_count_2}")

print(code)
```
