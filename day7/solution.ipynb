```python
import numpy as np

data = []
with open('input.txt', 'r') as file:
    for line in file.readlines():
        data.append(list(line[:-1]))

data = np.array(data)
```

```python
def problem_1(data: np.ndarray) -> tuple[int, int]:
    m, n = data.shape
    splits = 0

    for i in range(1, m):
        if i % 2 != 0:
            data[i, data[i-1, :] == 'S'] = 'S'
        else:
            marked = np.where(np.logical_and(data[i, :] == '^', data[i-1, :] == 'S'))[0]
            for j in marked:
                data[i, j+1] = data[i, j-1] = 'S'
                splits += 1
            data[i, np.logical_and(data[i-1, :] == 'S', data[i, :] == '.')] = 'S'

    return (np.sum(data[-1, :] == 'S').item(), splits)

problem_1(data)
```

```python
from functools import lru_cache

def problem_2(data: np.ndarray) -> int:
    m, n = data.shape

    # solution by backtracking
    @lru_cache(maxsize=None)
    def backtrack(i: int, j: int) -> int:
        # travel downwards until splitter/bottom
        while i < m and data[i, j] == 'S':
            i += 1

        if i == m:
            return 1

        total = 0
        if j + 1 < n:
            total += backtrack(i, j + 1)
        if j - 1 >= 0:
            total += backtrack(i, j - 1)

        return total

    start = np.where(data[0, :] == 'S')[0][0]
    return backtrack(0, start)

problem_2(data)
```
