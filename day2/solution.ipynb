```python
import csv

reader = csv.reader(open('input.csv', 'r'))
ids = [[tuple(ele.split('-')) for ele in row] for row in reader][0]
```

```python
def invalid_ids(lower: str, upper: str) -> int:
    """
    Returns sum of any invalid ids in the range

    Invalid id form: <num><num> repeated exactly twice
    """
    m, n = len(lower), len(upper)
    if m == n and m%2 != 0:
        return 0
    
    lower, upper = int(lower), int(upper)
    
    tot = 0
    lower_power = m if m%2==0 else m+1
    upper_power = n if n%2==0 else n-1

    lower_lim = lower if lower_power == m else 10**(m)
    upper_lim = upper if upper_power == n else 10**(n)

    start = str(lower_lim)[:lower_power//2]
    current = int(start * 2)

    while current < lower_lim:
        start = str(int(start) + 1)
        current = int(start * 2)

    while current <= upper_lim:
        tot += current
        start = str(int(start) + 1)
        current = int(start * 2)

    return tot

ans = 0

for id in ids:
    tot = invalid_ids(*id)
    ans += tot

print(ans)

```

```python
import math

def split_in_intervals(s: str, step: int, n: int) -> bool:
    last = None
    for i in range(0, n, step):
        substr = s[i:i+step]
        if not last:
            last = substr
        elif substr != last:
            return False
        
    return True

def invalid_ids_2(lower: str, upper: str) -> int:
    """
    Now an invalid id must have at least 2 repeated blocks
    """
    # brute force approach
    tot = 0

    for num in range(int(lower), int(upper) + 1):
        n = math.ceil(math.log10(num))
        # doubles
        s = str(num)
        
        split = 1
        while split <= n // 2:
            if split_in_intervals(s, split, n):
                tot += num
                break

            split += 1

    return tot


ans = 0

for id in ids:
    tot = invalid_ids_2(*id)
    ans += tot

print(ans)
    
```
