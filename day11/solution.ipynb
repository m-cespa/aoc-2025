```python
graph = {}

with open('input.txt', 'r') as file:
    for line in file.readlines():
        codes = line[:-1].split(' ')
        graph[codes[0][:-1]] = codes[1:]
```

```python
from collections import deque

def paths_to_goal(root: str, goal: str) -> int:
    ans = 0

    queue = deque([root])

    while queue:
        node = queue.pop()

        if node == goal:
            ans += 1
            continue

        for child in graph[node]:
            queue.appendleft(child)

    return ans

paths_to_goal('you', 'out')
```

```python
from functools import lru_cache

def paths_through_nodes(root: str, goal: str, stops: list[str]) -> int:
    stop_idx = {stop: i for i, stop in enumerate(stops)}
    ALL_VISITED = (1 << len(stops)) - 1 # corresponds to all bits = 1

    @lru_cache(maxsize=None)
    def dfs(node: str, visited: int) -> int:
        if node in stop_idx:
            visited |= (1 << stop_idx[node]) # bitwise OR turns on bits for visited nodes

        if node == goal:
            return 1 if visited == ALL_VISITED else 0
        
        total = 0

        for child in graph[node]:
            total += dfs(child, visited)

        return total
    
    return dfs(root, 0)
    
paths_through_nodes('svr', 'out', ['dac', 'fft'])
        

```
