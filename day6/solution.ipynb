```python
import numpy as np

data = np.genfromtxt('input.txt', dtype=int, skip_footer=1)
operations = np.genfromtxt('input.txt', dtype=str, skip_header=4)
```

```python
def compute(data: np.ndarray, operations: np.ndarray) -> int:
    ans = 0

    for i, op in enumerate(operations):
        if op == '*':
            ans += np.prod(data[:, i])
        elif op == '+':
            ans += np.sum(data[:, i])

    return int(ans)

compute(data, operations)
```

```python
def preprocessor(path: str) -> list[np.ndarray]:
    lines = []

    with open(path, 'r') as file:
        for line in file.readlines()[:-1]:
            lines.append(line[:-1]) # remove '\n' chara

    n = len(lines) # number of operation rows

    def gen_matrix(n: int) -> list[list[str]]:
        return [['0'] for _ in range(n)]
    
    matrix = gen_matrix(n)
    processed = []

    for charas in zip(*lines):
        if all(c==' ' for c in charas):
            processed.append(np.array(matrix, dtype=str)[:, 1:])
            matrix = gen_matrix(n)
            continue

        for i, c in enumerate(charas):
            matrix[i].append(c if c!=' ' else '0')

    # append final matrix
    processed.append(np.array(matrix, dtype=str)[:, 1:])

    return processed

def compute_2(data: list[np.ndarray], operations: np.ndarray) -> int:
    ans = 0

    for i, op in enumerate(operations):
        mat = data[i].T
        nums = [eval(''.join(row).strip('0')) for row in mat]

        if op == '+':
            ans += np.sum(nums)
        elif op == '*':
            ans += np.prod(nums)

    return int(ans)

compute_2(preprocessor('input.txt'), operations)
```
