```python
import numpy as np
from collections import defaultdict

shapes = []

def converter(shape: list[str]) -> np.ndarray:
    n = len(shape)
    x = np.zeros((n, n), dtype=int)
    for i in range(n):
        for j in range(n):
            x[i, j] = 1 if shape[i][j] == '#' else 0

    return x

with open('shapes.txt', 'r') as file:
    lines = [line[:-1] for line in file.readlines()]

    for i in range(0, len(lines), 5):
        shapes.append(converter(lines[i+1:i+4]))

grids = defaultdict(list)

with open('grid.txt', 'r') as file:
    for line in file.readlines():
        blocks = line[:-1].split(' ')
        size = blocks[0][:-1].split('x')
        m, n = int(size[0]), int(size[1])


        grids[(m, n)].append(list(map(int, blocks[1:])))
```

```python
def naive_solve(grid_shape: tuple[int], counts: list[int]) -> bool:
    return np.prod(grid_shape) // 9 >= np.sum(counts)

ans = 0
for grid_shape in grids:
    for block in grids[grid_shape]:
        ans += naive_solve(grid_shape, block)

print(ans)
```
