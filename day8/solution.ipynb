```python
import numpy as np

data = np.genfromtxt("input.txt", delimiter=",", dtype=int)
```

```python
def problem_1(data: np.ndarray) -> int:
    # find 1000 shortest euclidean distances
    # count sizes of 3 largest groups
    m = len(data)

    # pairwise distances sorted
    pairs = sorted(
        [(i, j) for i in range(m) for j in range(i + 1, m)],
        key=lambda x: np.linalg.norm(data[x[0]] - data[x[1]]), # type: ignore
    ) # type: ignore

    # map connections as an undirected graph
    graph = {}
    for a, b in pairs[:1000]: # only connect closest 1000 pairs
        if a in graph:
            graph[a].append(b)
        else:
            graph[a] = [b]
        if b in graph:
            graph[b].append(a)
        else:
            graph[b] = [a]

    # traverse and count size of largest disconnected islands
    visited = set()
    sizes = []

    def dfs(node: int) -> int:
        if node in visited:
            return 0
        
        total = 1
        visited.add(node)
        for nxt in graph[node]:
            total += dfs(nxt)

        return total

    for start in graph:
        sizes.append(dfs(start))

    return int(np.prod(sorted(sizes, reverse=True)[:3]))

problem_1(data)
```

```python
def problem_2(data: np.ndarray) -> int:
    # find 1000 shortest euclidean distances
    # count sizes of 3 largest groups
    m = len(data)

    # pairwise distances sorted
    pairs = sorted(
        [(i, j) for i in range(m) for j in range(i + 1, m)],
        key=lambda x: np.linalg.norm(data[x[0]] - data[x[1]]), # type: ignore
    ) # type: ignore

    # full set
    to_visit = set(range(0, 1000))
    top_idx = None

    # keep expanding pair choices until it is one connected graph
    for i in range(1, 499500):
        # map connections as an undirected graph
        graph = {}
        for a, b in pairs[:i]: # only connect closest 1000 pairs
            if a in graph:
                graph[a].append(b)
            else:
                graph[a] = [b]
            if b in graph:
                graph[b].append(a)
            else:
                graph[b] = [a]

        # traverse and count size of largest disconnected islands
        visited = set()

        def dfs(node: int) -> None:
            if node in visited:
                return
            
            visited.add(node)
            for nxt in graph[node]:
                dfs(nxt)

        dfs(pairs[0][0])
        if visited == to_visit:
            top_idx = i - 1
            break

    if top_idx is None:
        raise ValueError

    return data[pairs[top_idx][0], 0] * data[pairs[top_idx][1], 0]

problem_2(data)
```
